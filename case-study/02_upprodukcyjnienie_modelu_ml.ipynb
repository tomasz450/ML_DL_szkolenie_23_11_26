# Jak uprodukcyjnic model z Jupyter Notebooka


## Co trzeba zrobic

1. Zamien notebook na skrypt/projekt (`train`, `predict`, `api`).
   Przykladowo: Python, scikit-learn Pipeline, FastAPI.
2. Zapisz model i preprocessing jako jeden artefakt.
   Przykladowo: joblib/pickle, MLflow Model Registry.
3. Wystaw predykcje przez API (`/health`, `/predict`).
   Przykladowo: FastAPI lub Flask.
4. Spakuj aplikacje do Dockera.
   Przykladowo: Docker, Docker Compose.
5. Wdroz na srodowisko (np. Kubernetes).
   Przykladowo: Kubernetes (lub AKS)
6. Dodaj CI/CD (testy -> build -> deploy).
   Przykladowo: GitHub Actions
7. Wlacz monitoring (bledy, opoznienia, jakosc modelu).
   Przykladowo: Prometheus + Grafana, MLflow, Langfuse (dla warstwy LLM).
8. Przygotuj plan rollback i aktualizacji modelu.
   Przykladowo: wersjonowanie modelu w MLflow, update na Kubernetes.
