### Jaki problem rozwiazujemy?
To problem klasyfikacji binarnej:
- klasa pozytywna: Loan_Status = Y (kredyt przyznany),
- klasa negatywna: Loan_Status = N (kredyt odrzucony).

Celem jest zbudowanie modelu, ktory jak najlepiej przewidzi decyzje kredytowa na podstawie dostepnych cech klienta i wniosku.

### Co oznaczaja kolumny?
- Loan_ID: unikalny identyfikator wniosku.
- Gender: plec wnioskodawcy.
- Married: stan cywilny (czy osoba jest w zwiazku malzenskim).
- Dependents: liczba osob na utrzymaniu.
- Education: poziom wyksztalcenia.
- Self_Employed: informacja, czy osoba jest samozatrudniona.
- ApplicantIncome: dochod wnioskodawcy.
- CoapplicantIncome: dochod wspolwnioskodawcy.
- LoanAmount: kwota wnioskowanego kredytu.
- Loan_Amount_Term: okres kredytowania (zwykle w miesiacach).
- Credit_History: historia kredytowa (uproszczony wskaznik wiarygodnosci).
- Property_Area: typ lokalizacji nieruchomosci (np. miejska, podmiejska, wiejska).
- Loan_Status: zmienna docelowa (Y - kredyt przyznany, N - kredyt odrzucony).

```python
from sklearn.metrics import accuracy_score

# Przyklad uzycia metryki Accuracy
y_true = [1, 0, 1, 1, 0, 1]  # Prawdziwe etykiety
y_pred = [1, 0, 0, 1, 0, 1]  # Predykcje modelu

acc = accuracy_score(y_true, y_pred)
print(f"Accuracy: {acc:.4f}")
```
