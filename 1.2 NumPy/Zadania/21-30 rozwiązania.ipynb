### Ćwiczenie 25
Podany jest wektor `v = np.array([1, 2, 3, 4])` (kształt _(4,)_). Zamień go na:

a) dwuwymiarowy __wiersz__ o kształcie _(1, 4)_,

b) dwuwymiarową __kolumnę__ o kształcie _(4, 1)_.

__Oczekiwany wynik:__

a)

```
array([[1, 2, 3, 4]])
```

b)

```
array([[1],
       [2],
       [3],
       [4]])
```

__Wskazówka:__ Użyj metody _reshape()_ lub `np.newaxis`.

````python
# b)
v.reshape(-1, 1)   # lub v[:, np.newaxis]
````

### Ćwiczenie 26
Podane są dwie tablice $A$ i $B$:

```
A = np.array([[3, 4, 5],
              [8, 3, 1]])
B = np.array([[0, 5, 2],
              [4, 2, 1]])
```

Połącz je tak, aby $B$ znalazła się __pod__ $A$ (wynik ma 4 wiersze i 3 kolumny).

__Oczekiwany wynik:__

```
array([[3, 4, 5],
       [8, 3, 1],
       [0, 5, 2],
       [4, 2, 1]])
```

__Wskazówka:__ Użyj funkcji _np.concatenate()_ z parametrem _axis_ (lub _np.vstack()_).

````python
A = np.array([[3, 4, 5],
              [8, 3, 1]])
B = np.array([[0, 5, 2],
              [4, 2, 1]])
````

````python
np.concatenate((A, B), axis=0)
````

### Ćwiczenie 27
Dla tablic $A$ i $B$ z poprzedniego ćwiczenia połącz je tak, aby $B$ znalazła się __obok__ $A$ (wynik ma 2 wiersze i 6 kolumn).

__Oczekiwany wynik:__

```
array([[3, 4, 5, 0, 5, 2],
       [8, 3, 1, 4, 2, 1]])
```

__Wskazówka:__ Użyj funkcji _np.concatenate()_ z parametrem _axis_ (lub _np.hstack()_).

````python
A = np.array([[3, 4, 5],
              [8, 3, 1]])
B = np.array([[0, 5, 2],
              [4, 2, 1]])
````

````python
np.concatenate((A, B), axis=1)
````

### Ćwiczenie 28

data = np.array([[4.3, 4.2],
                 [3.1, 3.6]])
target = np.array([[0],

Połącz je w jedną tablicę o kształcie _(2, 3)_, w której _target_ jest ostatnią kolumną.

### Ćwiczenie 29
Podane są trzy jednowymiarowe tablice (trzy cechy trzech obserwacji):

```
feature1 = np.array([1.6, 0.9, 2.2])
feature2 = np.array([0.4, 1.3, 3.2])
feature3 = np.array([1.4, 0.3, 1.2])
```

Sklej je w jedną tablicę o kształcie _(3, 3)_, w której każda z tych tablic jest __kolumną__.

__Oczekiwany wynik:__

```
array([[1.6, 0.4, 1.4],
       [0.9, 1.3, 0.3],
       [2.2, 3.2, 1.2]])
```

__Wskazówka:__ Użyj funkcji _np.column_stack()_.

````python
feature1 = np.array([1.6, 0.9, 2.2])
feature2 = np.array([0.4, 1.3, 3.2])
feature3 = np.array([1.4, 0.3, 1.2])
````

````python
# Rozwiązanie 1
np.column_stack((feature1, feature2, feature3))
````

````python
# Rozwiązanie 2: zamiana wektorów na kolumny i sklejenie wzdłuż osi 1
np.hstack([f.reshape(-1, 1) for f in (feature1, feature2, feature3)])
````

### Ćwiczenie 30
Podana jest tablica `X` o kształcie _(6, 4)_ - 6 obserwacji, każda z 4 cechami:

```
X = np.arange(24).reshape(6, 4)
```

Podziel ją na zbiór treningowy (pierwsze 4 wiersze) i zbiór testowy (ostatnie 2 wiersze), a następnie wypisz kształty obu zbiorów.

__Oczekiwany wynik:__

```
(4, 4) (2, 4)
```

__Wskazówka:__ Użyj wycinków `X[:4]` i `X[4:]` (lub funkcji _np.split()_).

````python
X = np.arange(24).reshape(6, 4)
````

````python
X_train = X[:4]
X_test = X[4:]
print(X_train.shape, X_test.shape)
````

````python
# Inny sposób
X_train, X_test = np.split(X, [4])
print(X_train.shape, X_test.shape)
````
