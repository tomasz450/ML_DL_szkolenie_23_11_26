## 1. Definicja uczenia maszynowego

**Uczenie maszynowe** (*Machine Learning*, ML) to dziedzina sztucznej inteligencji zajmująca się tworzeniem systemów, które **uczą się na podstawie danych**, bez konieczności jawnego programowania każdej możliwej sytuacji.

**Kluczowe elementy definicji:**
- **Uczenie się** – poprawa wydajności wraz z doświadczeniem.
- **Dane** – źródło informacji, na podstawie których model się uczy.
- **Automatyzacja** – mniej ręcznego pisania reguł.
- **Adaptacja** – dostosowanie do nowych, niewidzianych wcześniej danych.

**Formalna definicja Toma Mitchella (1997):**

> Program komputerowy uczy się na podstawie doświadczenia *E* w odniesieniu do pewnej klasy zadań *T* i miary wydajności *P*, jeśli jego wydajność w zadaniach *T*, mierzona przez *P*, poprawia się wraz z doświadczeniem *E*.

*Przykład:* filtr spamu – zadanie **T**: klasyfikacja wiadomości, doświadczenie **E**: wiadomości oznaczone przez użytkowników, miara **P**: odsetek poprawnie sklasyfikowanych wiadomości.

## 3. Komponenty systemu i cykl życia projektu

**Komponenty systemu uczenia maszynowego:**
- **Dane wejściowe** – cechy opisujące problem.
- **Model** – matematyczna reprezentacja problemu.
- **Funkcja celu (straty)** – miara jakości predykcji modelu.
- **Algorytm optymalizacji** – metoda dostrajania parametrów modelu.

**Cykl życia projektu:** zbieranie i przygotowanie danych → trenowanie modelu → ewaluacja → wdrożenie → monitorowanie i aktualizacja. Szczegółowo omawiamy go w notebooku *1.4 CRISP-DM* i w `09_cykl_zycia_projektu_ML`.

![Data Science Life Cycle](../obrazki/2.3_data_science_life_cycle.png)

## 4. Rodzaje problemów rozwiązywanych przez uczenie maszynowe

- **Klasyfikacja** – przypisywanie etykiet do danych wejściowych (np. spam / nie spam).
- **Regresja** – przewidywanie wartości ciągłych (np. cena mieszkania).
- **Grupowanie (clustering)** – odkrywanie naturalnych grup w danych (np. segmentacja klientów).
- **Redukcja wymiarowości** – zmniejszanie liczby cech przy zachowaniu istotnych informacji.

**Główne paradygmaty uczenia** (szczegóły w notebooku `07_paradygmaty_uczenia_maszynowego`):
- **nadzorowane** – model uczy się na danych z etykietami (najczęściej używane),
- **nienadzorowane** – model szuka wzorców w danych bez etykiet,
- **ze wzmocnieniem** – model uczy się przez interakcję ze środowiskiem,
- **półnadzorowane** – mało danych z etykietami i dużo bez etykiet.

## 5. Uczenie maszynowe a sztuczna inteligencja i deep learning

- Uczenie maszynowe jest **podzbiorem** sztucznej inteligencji (AI), a AI obejmuje szerszy zakres technik symulowania inteligencji.
- Deep learning jest **podzbiorem** uczenia maszynowego: wykorzystuje głębokie sieci neuronowe do automatycznego uczenia reprezentacji danych.

## 7. Historia i ewolucja uczenia maszynowego

Najważniejsze kamienie milowe:

- **1943** – McCulloch i Pitts tworzą pierwszy model sztucznego neuronu.
- **1950** – Alan Turing proponuje test Turinga.
- **1957** – Frank Rosenblatt tworzy perceptron, pierwszy trenowalny model neuronowy.
- **1969** – książka „Perceptrons” pokazuje ograniczenia perceptronów: początek **pierwszej „zimy AI”**.
- **1986** – Rumelhart, Hinton i Williams popularyzują wsteczną propagację błędu.
- **1992–1995** – powstają maszyny wektorów nośnych (SVM).
- **1997** – IBM Deep Blue pokonuje Garry'ego Kasparowa w szachach.
- **1998** – Yann LeCun przedstawia sieć konwolucyjną LeNet-5 (rozpoznawanie pisma ręcznego).
- **2006** – Hinton i współpracownicy pokazują skuteczny trening głębokich sieci: początek ery deep learningu.
- **2012** – AlexNet wygrywa konkurs ImageNet; deep learning zaczyna dominować w rozpoznawaniu obrazów.
- **2014** – Ian Goodfellow wprowadza sieci GAN.
- **2016** – AlphaGo pokonuje mistrza świata w Go, Lee Sedola.
- **2017** – architektura Transformer rewolucjonizuje przetwarzanie języka naturalnego.
- **2020** – OpenAI prezentuje GPT-3.
- **2022** – ChatGPT upowszechnia generatywną sztuczną inteligencję.
- **2024** – **Nagroda Nobla z fizyki** dla Johna J. Hopfielda i Geoffreya E. Hintona „za fundamentalne odkrycia i wynalazki, umożliwiające uczenie maszynowe z wykorzystaniem sztucznych sieci neuronowych”.

![Nagroda Nobla z fizyki 2024](../obrazki/2.3_nobel_2024.png)

## 8. Wpływ mocy obliczeniowej i dostępności danych

Rozwój uczenia maszynowego był ściśle powiązany z postępem sprzętu i gromadzeniem danych:

- **Moc obliczeniowa:** procesory wielordzeniowe, karty graficzne (GPU), wyspecjalizowane układy (np. Google TPU) i chmury obliczeniowe dające dostęp do dużych zasobów na żądanie.
- **Dane:** cyfryzacja, Internet i media społecznościowe, Internet Rzeczy (IoT), technologie Big Data, otwarte dane.
- **Skutki:** możliwość trenowania coraz większych modeli, szybsze eksperymenty, gotowe narzędzia i modele dostępne dla każdego (np. transfer learning).
- **Koszty:** prywatność danych, interpretowalność złożonych modeli, zużycie energii i nierówny dostęp do zasobów.
