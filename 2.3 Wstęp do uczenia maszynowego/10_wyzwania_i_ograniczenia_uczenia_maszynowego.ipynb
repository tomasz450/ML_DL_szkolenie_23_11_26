# Wyzwania i ograniczenia uczenia maszynowego

Uczenie maszynowe daje świetne wyniki, ale ma też realne ograniczenia. Najważniejsze to:

## 1. Dane: jakość i reprezentatywność
**Problem:** słabe dane = słaby model (*garbage in, garbage out*).

Najczęstsze trudności:
- braki danych,
- błędne lub zaszumione dane,
- niezbalansowane klasy,
- stronniczość danych,
- zmiana rozkładu danych w czasie (*concept drift*).

Co pomaga:
- porządne czyszczenie danych i EDA,
- balansowanie klas (np. oversampling / SMOTE),
- monitoring jakości danych po wdrożeniu,
- regularne ponowne trenowanie modelu.

---

## 2. Koszt obliczeń i infrastruktura
**Problem:** dobre modele często wymagają dużo czasu i mocy obliczeniowej.

Objawy:
- długie treningi,
- wysokie koszty GPU / CPU,
- trudności z wdrożeniem na słabszych urządzeniach.

Co pomaga:
- zacząć od prostego baseline,
- używać transfer learningu,
- ograniczać złożoność modelu,
- kompresja modeli (np. kwantyzacja),
- dobór rozwiązania do realnych ograniczeń projektu.

---

## 3. Generalizacja modelu
**Problem:** model może działać świetnie na treningu, ale słabo na nowych danych (*overfitting*) albo być zbyt prosty, by uchwycić wzorce (*underfitting*).

Co pomaga: poprawny podział danych, walidacja krzyżowa, regularyzacja i analiza błędów zamiast patrzenia na jedną metrykę. Szczegóły w notebooku `06_podstawowe_koncepcje`.

---

## 4. Interpretowalność i zaufanie
**Problem:** niektóre modele są skuteczne, ale trudne do wyjaśnienia.

Dlaczego to ważne:
- w medycynie, finansach i prawie trzeba uzasadniać decyzje,
- użytkownik musi ufać wynikowi modelu.

Co pomaga:
- wybór prostszych modeli tam, gdzie to możliwe,
- narzędzia wyjaśnialności (np. SHAP, LIME),
- jasne raportowanie ograniczeń modelu.

---

## 5. Bezpieczeństwo, etyka i zgodność
**Problem:** modele mogą dziedziczyć bias, być podatne na manipulacje i naruszać prywatność.

Ryzyka:
- dyskryminacja wybranych grup,
- ataki adversarial,
- naruszenia prywatności danych.

Co pomaga:
- testy fairness,
- anonimizacja i minimalizacja danych,
- kontrola dostępu i audyt modeli,
- zgodność z regulacjami (np. RODO, AI Act).
