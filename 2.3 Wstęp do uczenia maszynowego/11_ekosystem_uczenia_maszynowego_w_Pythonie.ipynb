# Ekosystem uczenia maszynowego w Pythonie

![Ekosystem ML w Pythonie](../obrazki/2.3_ekosystem_ml.png)

Grafika pokazuje także narzędzia, które straciły na znaczeniu: **Theano** (rozwój zakończono w 2017 r.) i **MXNet** (projekt zarchiwizowany w 2023 r.). W praktyce dominują dziś **PyTorch** i **TensorFlow/Keras**.

Python jest popularny w ML, bo ma prosty i spójny ekosystem narzędzi. Najlepiej myśleć o nim jak o potoku (*pipeline*): dane → analiza → model → wdrożenie.

**Plan notatnika:**
1. Praca z danymi
2. Wizualizacja
3. Modelowanie ML/DL
4. NLP
5. Wdrożenie
6. Minimalny stack na start

## 1. Praca z danymi
- **NumPy** – szybkie operacje numeryczne na tablicach.
- **Pandas** – tabele (DataFrame), czyszczenie i transformacje danych.
- **SciPy** – dodatkowe metody matematyczne i statystyczne.

*Kiedy używać:* prawie w każdym projekcie startujesz od NumPy + Pandas.

## 2. Wizualizacja
- **Matplotlib** – podstawowe wykresy.
- **Seaborn** – ładniejsze wykresy statystyczne.
- **Plotly** – wykresy interaktywne (np. dashboardy).

*Kiedy używać:* EDA, prezentacja wyników i analiza błędów modelu.

## 3. Modelowanie ML/DL
- **Scikit-learn** – klasyczne ML (regresja, klasyfikacja, grupowanie).
- **XGBoost, LightGBM** – gradient boosting, bardzo skuteczny dla danych tabelarycznych.
- **TensorFlow / Keras** – deep learning, szybkie budowanie sieci (Keras 3 działa na TensorFlow, PyTorch lub JAX).
- **PyTorch** – deep learning, większa elastyczność i popularność w badaniach.

*Kiedy używać:*
- **Scikit-learn**: pierwszy wybór dla większości problemów tabelarycznych,
- **TensorFlow / PyTorch**: gdy potrzebujesz sieci neuronowych lub dużych modeli.

## 4. NLP
- **NLTK** – edukacja i klasyczne przetwarzanie tekstu.
- **spaCy** – szybkie przetwarzanie produkcyjne.
- **Transformers (Hugging Face)** – nowoczesne modele językowe (BERT, GPT).

*Kiedy używać:*
- klasyczne zadania tekstowe → spaCy / NLTK,
- zaawansowane zadania językowe → Transformers.

## 5. Wdrożenie
- **Flask / FastAPI** – wystawienie modelu jako API.
- **Django** – większa aplikacja webowa z backendem.
- **Docker** – spakowanie aplikacji i łatwe wdrożenie.

*Kiedy używać:* model ma działać poza notebookiem i obsługiwać realne zapytania.
