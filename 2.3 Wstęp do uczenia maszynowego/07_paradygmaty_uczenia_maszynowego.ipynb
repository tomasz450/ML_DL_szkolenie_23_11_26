# Paradygmaty uczenia maszynowego

Najprościej: to 4 główne sposoby uczenia modeli. Różnią się tym, jakie dane mamy i jaki jest cel.

![Podział uczenia maszynowego](../obrazki/2.3_podzial_ml.png)

**Plan notatnika:**
1. Uczenie nadzorowane
2. Uczenie nienadzorowane
3. Uczenie ze wzmocnieniem
4. Uczenie półnadzorowane
5. Jak wybrać paradygmat?

## 1. Uczenie nadzorowane (Supervised Learning)

**Masz:** cechy + poprawne etykiety.

**Cel:** przewidzieć etykietę dla nowych danych.

**Typowe zadania:**
- klasyfikacja (np. spam / nie spam),
- regresja (np. cena mieszkania).

**Przykładowe algorytmy:** regresja liniowa, drzewa decyzyjne / las losowy, SVM, sieci neuronowe.

**Plusy:**
- zwykle dobra skuteczność,
- łatwo policzyć metryki jakości.

**Minusy:**
- potrzeba dużo dobrze opisanych (etykietowanych) danych.

---

## 2. Uczenie nienadzorowane (Unsupervised Learning)

**Masz:** tylko cechy, bez etykiet.

**Cel:** odkryć strukturę lub wzorce w danych.

**Typowe zadania:**
- segmentacja klientów,
- wykrywanie anomalii,
- redukcja wymiarowości.

**Przykładowe algorytmy:** K-means, DBSCAN, PCA, autoenkodery.

**Plusy:**
- działa bez etykiet,
- pomaga zrozumieć dane przed modelowaniem.

**Minusy:**
- trudniej ocenić jakość wyników,
- interpretacja bywa nieoczywista.

---

## 3. Uczenie ze wzmocnieniem (Reinforcement Learning)

**Masz:** agenta, środowisko i system nagród / kar.

**Cel:** nauczyć strategii działań, która maksymalizuje nagrodę w czasie.

**Typowe zadania:**
- gry,
- sterowanie robotem,
- optymalizacja decyzji sekwencyjnych.

**Przykładowe algorytmy:** Q-learning, DQN, PPO.

**Plusy:**
- dobre do problemów decyzyjnych krok po kroku.

**Minusy:**
- trening bywa długi i kosztowny,
- trudny balans: eksploracja vs eksploatacja.

---

## 4. Uczenie półnadzorowane (Semi-supervised Learning)

**Masz:** mało danych z etykietą + dużo danych bez etykiety.

**Cel:** poprawić model nadzorowany dzięki dodatkowym nieoznakowanym danym.

**Typowe zadania:**
- klasyfikacja tekstu i obrazów, gdy etykietowanie jest drogie.

**Plusy:**
- lepsze wyniki niż przy samym małym zbiorze etykietowanym,
- mniejszy koszt etykietowania.

**Minusy:**
- ryzyko propagacji błędnych pseudo-etykiet.

---

**Uwaga – uczenie samonadzorowane (self-supervised):** etykiety powstają automatycznie z samych danych (np. model językowy przewiduje następne słowo w tekście). Tak uczone są współczesne duże modele językowe (LLM), które poznamy na końcu kursu.

---

## 5. Jak wybrać paradygmat?
1. Jeśli masz dużo etykiet: zacznij od nadzorowanego.
2. Jeśli nie masz etykiet: zacznij od nienadzorowanego.
3. Jeśli problem to sekwencja decyzji: rozważ uczenie ze wzmocnieniem.
4. Jeśli etykiet jest mało, ale danych dużo: półnadzorowane.
