## 1. Rola danych w uczeniu maszynowym

Dane to najważniejszy element każdego projektu ML. Można mieć świetny algorytm, ale bez dobrych danych model nie będzie działał dobrze.

### Co jest najważniejsze?

**1. Jakość danych**
- brakujące wartości, błędy i szum pogarszają wyniki,
- czasem mniejszy, ale czysty zbiór jest lepszy niż duży i słaby jakościowo.

**2. Reprezentatywność**
- dane treningowe powinny przypominać realne przypadki,
- niezbalansowane klasy mogą faworyzować jedną grupę i psuć predykcje.

**3. Podział danych**
- dane dzielimy na zbiór treningowy, walidacyjny i testowy (szczegóły w notebooku `06_podstawowe_koncepcje`).

**4. Przygotowanie danych (preprocessing)**
- czyszczenie braków i błędów,
- skalowanie / standaryzacja,
- kodowanie zmiennych kategorycznych,
- ewentualnie tworzenie nowych cech (*feature engineering*).

**5. Dane zmieniają się w czasie**
- w praktyce pojawia się **concept drift** (zmiana zależności w danych),
- model trzeba monitorować i okresowo douczać.

### Typy danych w ML
- **Strukturalne**: tabele, liczby, rekordy (najprostsze na start).
- **Niestrukturalne**: tekst, obrazy, audio, wideo (wymagają dodatkowych technik).

### Aspekty praktyczne
- Dbaj o prywatność i zgodność z przepisami (np. RODO).
- Sprawdzaj stronniczość danych, żeby ograniczać ryzyko dyskryminacji.
- Przed modelowaniem zawsze zrób krótką EDA (rozkłady, braki, zależności).

## 2. Modele i algorytmy

### Czym jest model?
Model to matematyczny zapis zależności w danych. Po treningu potrafi przewidywać wyniki dla nowych przykładów.

Przykłady modeli: regresja liniowa, drzewo decyzyjne, sieć neuronowa.

### Czym jest algorytm?
Algorytm to sposób, w jaki model uczy się z danych (czyli jak aktualizuje swoje parametry).

### Jak wygląda uczenie modelu?
1. Wybierasz model i jego ustawienia (hiperparametry).
2. Trenujesz model na zbiorze treningowym.
3. Sprawdzasz wynik na walidacji / testach.
4. Poprawiasz ustawienia i powtarzasz proces.

### Parametry i hiperparametry
- **Parametry**: model uczy ich sam z danych (np. wagi).
- **Hiperparametry**: ustawiasz je przed treningiem (np. głębokość drzewa). Szczegóły w następnym notebooku.

### Co wybrać na początku?
1. Najpierw prosty i interpretowalny model.
2. Potem dopiero bardziej złożone modele.
3. Zawsze porównuj kilka modeli na tej samej metryce.
