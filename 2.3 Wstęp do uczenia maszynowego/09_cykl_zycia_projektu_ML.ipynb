## 2. Cykl życia projektu ML – wersja praktyczna

**1. Zdefiniuj problem biznesowy**
- co przewidujesz,
- jaką decyzję to wspiera,
- jaka metryka oznacza sukces.

**2. Przygotuj dane**
- zbierz źródła,
- sprawdź jakość,
- oczyść i przekształć dane,
- zrób sensowny podział train / validation / test.

**3. Zbuduj baseline i rozwijaj model**
- zacznij od prostego modelu bazowego,
- potem testuj mocniejsze algorytmy,
- strojenie rób dopiero po baseline.

**4. Ewaluacja i analiza błędów**
- oceniaj na zbiorze testowym,
- sprawdź, gdzie model się myli,
- zweryfikuj wynik pod kątem celu biznesowego.

**5. Wdrożenie i monitoring**
- wdrożenie wsadowe (batch) lub online,
- monitoruj metryki, jakość danych i drift,
- zaplanuj ponowne trenowanie i wersjonowanie modelu.

Najczęstsze pułapki (brak metryki, data leakage, pominięcie baseline) opisaliśmy w notebooku *1.4 CRISP-DM*. Dodatkowo: **brak monitoringu po wdrożeniu** – model starzeje się wraz ze zmianą danych.

## 3. Jak stawiać hipotezy w Data Science

Dobre hipotezy przyspieszają projekt i ograniczają losowe eksperymenty.

**1. Zacznij od rozmowy z biznesem**
- co ma się poprawić,
- jakie decyzje będą podejmowane na podstawie modelu,
- które przypadki są najtrudniejsze.

**2. Zrób szybki research**
- jak podobny problem rozwiązują inni,
- jakie podejścia zwykle działają,
- jakie są typowe pułapki.

**3. Sprawdź dane (mini-EDA)**
- braki danych,
- wartości odstające,
- korelacje,
- potencjalne źródła biasu.

**4. Zapisuj hipotezy w formie testowalnej**

Przykład: „Dodanie cechy X poprawi recall klasy 1 o co najmniej 3 p.p.”

**5. Pracuj iteracyjnie (agile)**
- testuj małymi krokami,
- regularnie pokazuj wyniki,
- aktualizuj hipotezy na podstawie feedbacku.

---

**Prosty szablon hipotezy:**
1. **Hipoteza:** co ma się poprawić?
2. **Zmiana:** co dokładnie zmieniam?
3. **Metryka:** czym mierzę efekt?
4. **Wynik:** poprawa / brak poprawy?
5. **Decyzja:** wdrażam czy odrzucam?
