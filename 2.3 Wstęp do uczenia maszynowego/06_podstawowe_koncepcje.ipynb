# Podstawowe koncepcje

3. Underfitting i overfitting

## 1. Cechy (features) i etykiety (labels)

To najważniejsze pojęcia w ML.

- **Cechy (X)** – informacje wejściowe, na podstawie których model podejmuje decyzję. Przykłady: metraż mieszkania, liczba pokoi, wiek klienta.
- **Etykieta (y)** – to, co model ma przewidzieć. Przykłady: cena mieszkania, spam / nie spam, rezygnacja klienta.

![Cechy i etykiety](../obrazki/2.3_cechy_i_etykiety.png)

**Jak to rozumieć w praktyce?**
- Model dostaje **cechy** i uczy się przewidywać **etykietę**.
- Jeśli etykieta jest błędna lub cechy są słabej jakości, model będzie działał źle.

**Najczęstsze typy zadań:**
1. **Klasyfikacja** – etykieta to klasa (np. 0/1, tak/nie).
2. **Regresja** – etykieta to liczba (np. cena, czas, sprzedaż).

Dobre cechy i poprawne etykiety mają często większy wpływ na wynik niż sam wybór algorytmu.

## 2. Zbiory treningowe, walidacyjne i testowe

Danych nie używamy „wszystkich naraz”. Dzielimy je na 2 lub 3 części:

1. **Treningowy (train)** – model uczy się wzorców; zwykle największa część danych.
2. **Walidacyjny (validation)** – porównywanie modeli i strojenie hiperparametrów; pomaga ograniczać przeuczenie.
3. **Testowy (test)** – używany tylko na końcu; pokazuje, jak model poradzi sobie na nowych danych.

Typowe proporcje to 60/20/20 lub 70/15/15, a przy dużych zbiorach nawet 98/1/1.

![Podział danych](../obrazki/2.3_podzial_danych.png)

![Train, validation, test](../obrazki/2.3_train_validation_test.png)

**Dobre praktyki:**
- Najpierw podziel dane, dopiero potem trenuj model.
- Dla klasyfikacji używaj podziału stratyfikowanego (zachowuje proporcje klas).
- Przy małych zbiorach korzystaj z walidacji krzyżowej (*cross-validation*) – rozdział 4.4.

## 3. Underfitting i overfitting

To dwa najczęstsze problemy podczas trenowania modeli.

### Underfitting (niedouczenie)
Model jest zbyt prosty i nie łapie wzorców.

*Objawy:* słaby wynik na treningu i słaby wynik na teście.

*Jak poprawić:*
- użyć bardziej elastycznego modelu,
- dodać lepsze cechy,
- trenować dłużej (jeśli ma to sens dla danego algorytmu).

### Overfitting (przeuczenie)
Model „zapamiętuje” trening zamiast uczyć się reguł ogólnych.

*Objawy:* bardzo dobry wynik na treningu i wyraźnie gorszy na teście.

*Jak poprawić:*
- uprościć model,
- dodać regularyzację,
- zebrać więcej danych,
- stosować walidację krzyżową lub *early stopping*.

### Cel w praktyce
Szukamy balansu: model ma być na tyle prosty, by nie przeuczać się, i na tyle mocny, by nie niedouczać się. Przykłady w kodzie – w rozdziale 4.3.
