# Zastosowania uczenia maszynowego

Poniżej skrócona mapa zastosowań. Każdy przykład opisany jest w tym samym schemacie:
**Problem → Co przewiduje model → Po co to robimy**.

## 1. Biznes i finanse
- **Ryzyko kredytowe** → prawdopodobieństwo spłaty → lepsze decyzje kredytowe.
- **Wykrywanie oszustw** → czy transakcja jest podejrzana → mniejsze straty finansowe.
- **Prognozowanie popytu / sprzedaży** → przyszła sprzedaż → lepsze planowanie budżetu i zapasów.
- **Automatyzacja obsługi klienta** → klasyfikacja zgłoszeń / intencji → szybsza obsługa.

## 2. Medycyna
- **Analiza obrazów medycznych** → obecność zmian chorobowych → szybsza diagnostyka.
- **Wsparcie decyzji terapeutycznych** → ryzyko powikłań / skuteczność terapii → bardziej spersonalizowane leczenie.
- **Monitoring pacjentów** → wykrywanie pogorszenia stanu → wcześniejsza reakcja personelu.

## 3. Przemysł i produkcja
- **Predykcyjne utrzymanie maszyn** → kiedy może wystąpić awaria → mniej przestojów.
- **Kontrola jakości (computer vision)** → czy produkt ma defekt → stabilna jakość.
- **Optymalizacja łańcucha dostaw** → popyt, czas dostawy, zapasy → niższe koszty operacyjne.

## 4. Marketing i e-commerce
- **Systemy rekomendacyjne** → co użytkownik może kupić → większa sprzedaż.
- **Segmentacja klientów** → grupy o podobnych zachowaniach → lepsze kampanie marketingowe.
- **Predykcja odejścia klienta (churn)** → kto może zrezygnować → działania retencyjne.
- **Dynamiczne ceny** → najlepsza cena w danym momencie → lepsza marża i konwersja.

## 5. Transport, robotyka i autonomiczne systemy
- **Percepcja otoczenia** → obiekty, pasy ruchu, przeszkody → bezpieczna nawigacja.
- **Planowanie trasy / decyzji** → najlepszy kolejny ruch → sprawniejsze sterowanie.
- **Roboty i drony** → kontrola ruchu i zadania autonomiczne → większa efektywność pracy.

## 6. Język i generatywna AI
- **Chatboty i asystenci** → odpowiedź na pytanie użytkownika → szybsza obsługa i dostęp do wiedzy.
- **Streszczanie i klasyfikacja tekstów** → najważniejsze informacje / kategoria dokumentu → oszczędność czasu analityków.
- **Asystenci programistów** → sugestie kodu → szybsze tworzenie oprogramowania.

---

## Co łączy większość tych zastosowań?
1. Dane historyczne + dane bieżące.
2. Model, który przewiduje klasę, liczbę lub ryzyko.
3. Decyzja biznesowa lub operacyjna oparta na wyniku modelu.

## Dobra praktyka
Przy każdym nowym case study zadaj 3 pytania:
1. **Co dokładnie przewiduję?** (klasa, liczba, prawdopodobieństwo)
2. **Jak zmierzę sukces?** (np. accuracy, recall, MAE)
3. **Jak ta predykcja pomoże w decyzji?** (konkretny efekt biznesowy)
