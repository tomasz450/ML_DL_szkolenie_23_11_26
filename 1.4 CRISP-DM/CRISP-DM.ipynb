# Wprowadzenie do CRISP-DM

Mamy już narzędzia (NumPy, statystyka, algebra), teraz potrzebujemy **schematu pracy**. **CRISP-DM** (*Cross-Industry Standard Process for Data Mining*) to najpopularniejsza metodyka projektów analitycznych i ML. Składa się z 6 faz i jest **iteracyjna** – po każdym etapie możemy wrócić do wcześniejszego.

![CRISP-DM](../obrazki/1.4_crisp_dm.png)

| Faza | Pytanie | Co robimy | Gdzie na szkoleniu |
|---|---|---|---|
| 1. Zrozumienie biznesu | Jaki problem rozwiązujemy i jak zmierzymy sukces? | cel, metryka sukcesu (KPI), ograniczenia | wstęp do każdego case study |
| 2. Zrozumienie danych | Jakie mamy dane i czy są dobre? | zbieranie, EDA, kontrola jakości (braki, błędy, outliery) | Dzień 2: Pandas, EDA |
| 3. Przygotowanie danych | Jak przygotować dane dla modelu? | czyszczenie, kodowanie kategorii, skalowanie, nowe cechy, podział train/test | Dzień 2: scikit-learn, feature engineering |
| 4. Modelowanie | Który model działa najlepiej? | trenowanie modeli, strojenie hiperparametrów | Dni 3–5: regresja, drzewa, lasy, sieci |
| 5. Ewaluacja | Czy model rozwiązuje problem biznesowy? | metryki na zbiorze testowym, analiza błędów | Dni 3–4: metryki, walidacja krzyżowa |
| 6. Wdrożenie | Jak dostarczyć wartość użytkownikom? | wdrożenie, monitoring, ponowne trenowanie | Dzień 5: aplikacja z modelem |

**Przykład – odejścia klientów (churn):**
1. *Biznes:* chcemy zatrzymać klientów; sukces = mniej odejść przy rozsądnym koszcie kampanii retencyjnej.
2. *Dane:* historia klientów z informacją, kto odszedł; sprawdzamy braki i rozkłady (np. wartości odstające i rozkłady z poprzedniego notatnika).
3. *Przygotowanie:* uzupełnienie braków, zakodowanie kategorii, skalowanie, podział na zbiór treningowy i testowy.
4. *Modelowanie:* regresja logistyczna, drzewo, las losowy – porównanie.
5. *Ewaluacja:* ile odejść model wychwytuje (recall), ile fałszywych alarmów daje (precision).
6. *Wdrożenie:* lista klientów zagrożonych odejściem co tydzień, monitoring jakości modelu.

### Trzy najczęstsze pułapki
- **Brak jasnej metryki sukcesu** – nie wiadomo, kiedy model jest „dobry”.
- **Data leakage** – do modelu wpadają informacje, których nie będzie w chwili predykcji (daje to zawyżone wyniki, które znikają po wdrożeniu).
- **Przeskoczenie do „zaawansowanego modelu”** bez prostego punktu odniesienia (*baseline*) – zawsze zaczynamy od prostego modelu i sprawdzamy, czy złożony faktycznie go pokonuje.

**Na kolejnych dniach** każde case study przechodzimy według tych 6 faz: od zrozumienia problemu, przez dane i model, po ocenę wyniku.
