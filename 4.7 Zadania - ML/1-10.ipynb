### Import bibliotek i dane

````python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer

data = {
    'size': ['XL', 'L', 'M', np.nan, 'M', 'M'],
    'color': ['red', 'green', 'blue', 'green', 'red', 'green'],
    'gender': ['female', 'male', np.nan, 'female', 'female', 'male'],
    'price': [199.0, 89.0, np.nan, 129.0, 79.0, 89.0],
    'weight': [500, 450, 300, np.nan, 410, np.nan],
    'bought': ['yes', 'no', 'yes', 'no', 'yes', 'no']
}

pd.DataFrame(data=data)
````

Za pomocą klasy `SimpleImputer` ze scikit-learn uzupełnij braki w kolumnie `weight` **średnią** wartością tej kolumny. Zmianę zapisz w `df` i wydrukuj cały obiekt `df`.


__Wskazówka:__ `SimpleImputer(strategy='mean')` oraz `fit_transform()`. Przekaż kolumnę w podwójnych nawiasach (`df[['weight']]`), bo imputer wymaga dwuwymiarowych danych.

### Ćwiczenie 7
Imputer po dopasowaniu zapamiętuje wartości użyte do uzupełnienia braków. Dopasuj `SimpleImputer` (strategia `'mean'`) do kolumny `weight` i wydrukuj **wartość średnią**, którą wstawiono w miejsce braków.

__Oczekiwany wynik:__

```
415.0
```

__Wskazówka:__ Atrybut `statistics_` zawiera wartość użytą dla każdej kolumny.

````python
df = pd.DataFrame(data=data)
df
````

````python
imputer = SimpleImputer(strategy='mean')
imputer.fit(df[['weight']])
print(imputer.statistics_[0])
````

Za pomocą `SimpleImputer` uzupełnij braki w kolumnie `price` **stałą wartością 99.0**. Zmianę zapisz w `df` i wydrukuj obiekt `df`.

### Ćwiczenie 10
Informacja o tym, że wartość **brakowała**, bywa sama w sobie cenną cechą. Uzupełnij kolumny `price` i `weight` **medianą** za pomocą `SimpleImputer` z argumentem `add_indicator=True` i wydrukuj otrzymaną macierz.

Macierz ma 4 kolumny: dwie uzupełnione oraz dwie z wskaźnikami braków (`1` tam, gdzie wartość uzupełniono).

__Oczekiwany wynik:__

```
[[199. 500.   0.   0.]
 [ 89. 450.   0.   0.]
 [ 89. 300.   1.   0.]
 [129. 430.   0.   1.]
 [ 79. 410.   0.   0.]
 [ 89. 430.   0.   1.]]
```

__Wskazówka:__ Parametr `add_indicator=True` dokleja do wyniku kolumny wskaźnikowe (0/1).

````python
df = pd.DataFrame(data=data)
df
````

````python
imputer = SimpleImputer(strategy='median', add_indicator=True)
result = imputer.fit_transform(df[['price', 'weight']])
print(result)
````
