### Ćwiczenie 17
Podano dane `X` (wzrost w cm) podzielone na zbiór treningowy i testowy.

a) Dopasuj `StandardScaler` **tylko na zbiorze treningowym**, a następnie przekształć zbiór treningowy i testowy.

b) Dla zbioru treningowego wydrukuj, czy średnia po skalowaniu wynosi 0 (`np.isclose`), oraz odchylenie standardowe (2 miejsca). Dla zbioru testowego wydrukuj średnią i odchylenie standardowe (2 miejsca). Dlaczego dla testowego nie wychodzi dokładnie 0 i 1?

__Oczekiwany wynik:__

```
treningowy: True 1.0
testowy: 0.16 0.96
```

__Wskazówka:__ Na zbiorze testowym wywołujemy tylko `transform()` (bez `fit`), żeby nie „podglądać” danych testowych.

````python
rng = np.random.RandomState(0)
X = rng.normal(loc=170, scale=10, size=(100, 1))
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
X_train.shape, X_test.shape
````

````python
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)
print('treningowy:', np.isclose(X_train_s.mean(), 0), X_train_s.std().round(2))
print('testowy:', X_test_s.mean().round(2), X_test_s.std().round(2))
````

### Ćwiczenie 18
W danych `df` są kolumny: `color` (kategoria), `size` (kategoria porządkowa) oraz `price` (liczba).

Za pomocą `ColumnTransformer` zbuduj jedno przekształcenie:
- `price` - `StandardScaler`,
- `color` - `OneHotEncoder(sparse_output=False)`,
- `size` - `OrdinalEncoder` z kolejnością `'S' < 'M' < 'L' < 'XL'`.

Wydrukuj nazwy kolumn wyniku (`get_feature_names_out()`) oraz kształt wyniku.

__Oczekiwany wynik:__

```
['num__price' 'nom__color_blue' 'nom__color_green' 'nom__color_red'
 'ord__size']
(6, 5)
```

__Wskazówka:__ Każdy element `ColumnTransformer` to krotka `(nazwa, transformer, kolumny)`.

````python
df = pd.DataFrame({
    'color': ['red', 'green', 'blue', 'green', 'red', 'green'],
    'size': ['XL', 'L', 'M', 'S', 'M', 'M'],
    'price': [199.0, 89.0, 150.0, 129.0, 79.0, 89.0],
})
df
````

````python
preprocessor = ColumnTransformer([
    ('num', StandardScaler(), ['price']),
    ('nom', OneHotEncoder(sparse_output=False), ['color']),
    ('ord', OrdinalEncoder(categories=[['S', 'M', 'L', 'XL']]), ['size']),
])
result = preprocessor.fit_transform(df)
print(preprocessor.get_feature_names_out())
print(result.shape)
````

### Ćwiczenie 19
Kolumna `date` zawiera daty. Utwórz z niej nowe cechy: `year` (rok), `month` (miesiąc), `dayofweek` (dzień tygodnia, 0 = poniedziałek) oraz `is_weekend` (1 dla soboty i niedzieli, w przeciwnym razie 0). Wydrukuj obiekt `dates`.

__Oczekiwany wynik:__

```
        date  year  month  dayofweek  is_weekend
0 2024-01-15  2024      1          0           0
1 2024-02-29  2024      2          3           0
2 2024-07-04  2024      7          3           0
3 2024-12-28  2024     12          5           1
4 2025-03-09  2025      3          6           1
```

__Wskazówka:__ Skorzystaj z akcesora `dt`: `dt.year`, `dt.month`, `dt.dayofweek`.

````python
dates = pd.DataFrame({
    'date': pd.to_datetime(['2024-01-15', '2024-02-29', '2024-07-04', '2024-12-28', '2025-03-09'])
})
dates
````

````python
dates['year'] = dates['date'].dt.year
dates['month'] = dates['date'].dt.month
dates['dayofweek'] = dates['date'].dt.dayofweek
dates['is_weekend'] = (dates['dayofweek'] >= 5).astype(int)
print(dates)
````

### Ćwiczenie 20
Kolumna `currency` zawiera **listy** walut akceptowanych w sklepie.

a) Dodaj kolumnę `number` z liczbą walut na liście.

b) Dodaj kolumnę `PLN_flag`, która ma wartość `1`, gdy na liście jest `'PLN'`, a w przeciwnym razie `0`.

Wydrukuj obiekt `df`.

__Oczekiwany wynik:__

```
               currency  number  PLN_flag
0            [PLN, USD]       2         1
1  [EUR, USD, PLN, CAD]       4         1
2                 [GBP]       1         0
3       [JPY, CZK, HUF]       3         0
4                    []       0         0
```

__Wskazówka:__ Użyj `apply()` z funkcją `len` oraz z funkcją anonimową (`lambda`) i operatorem `in`.

````python
df = pd.DataFrame({
    'currency': [
        ['PLN', 'USD'],
        ['EUR', 'USD', 'PLN', 'CAD'],
        ['GBP'],
        ['JPY', 'CZK', 'HUF'],
        [],
    ]
})
df
````

````python
df['number'] = df['currency'].apply(len)
df['PLN_flag'] = df['currency'].apply(lambda items: 1 if 'PLN' in items else 0)
print(df)
````
