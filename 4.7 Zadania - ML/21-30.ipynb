$$\theta = (X^T X)^{-1} X^T y$$

Zbuduj macierz $X$ z kolumną jedynek (wyraz wolny) i kolumną `x`. Wydrukuj parametry $\theta_0$ i $\theta_1$ zaokrąglone do 2 miejsc.

__Oczekiwany wynik:__

```
[3746.67  255.71]
```

__Wskazówka:__ Użyj `np.c_[]` do połączenia kolumn, `.T` do transpozycji, `@` do mnożenia macierzy i `np.linalg.inv()` do odwracania.

````python
x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([4000, 4300, 4450, 4800, 5000, 5300])
````

````python
X_b = np.c_[np.ones(len(x)), x]
theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
print(theta.round(2))
````

### Ćwiczenie 23
Dla rzeczywistych wartości `y_true` i predykcji `y_pred` oblicz **ręcznie** (bez `sklearn.metrics`):

$$\text{MAE} = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y}_i| \qquad \text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2 \qquad \text{RMSE} = \sqrt{\text{MSE}}$$

Wydrukuj trzy wartości zaokrąglone do 2 miejsc.

__Oczekiwany wynik:__

```
15.0 245.0 15.65
```

__Wskazówka:__ Działania na tablicach NumPy są wykonywane element po elemencie: `np.abs()`, `** 2`, `.mean()`, `np.sqrt()`.

````python
y_true = np.array([200, 250, 300, 350, 400])
y_pred = np.array([210, 240, 320, 330, 415])
````

````python
errors = y_true - y_pred
mae = np.abs(errors).mean()
mse = (errors ** 2).mean()
rmse = np.sqrt(mse)
print(round(mae, 2), round(mse, 2), round(rmse, 2))
````

### Ćwiczenie 24
Dla tych samych `y_true` i `y_pred` oblicz **współczynnik determinacji** $R^2$:

$$R^2 = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}$$

Policz go ręcznie, a następnie sprawdź wynik funkcją `r2_score()`. Wydrukuj obie wartości zaokrąglone do 3 miejsc.

__Oczekiwany wynik:__

```
0.951 0.951
```

__Wskazówka:__ Licznik to suma kwadratów reszt, mianownik - suma kwadratów odchyleń od średniej.

````python
y_true = np.array([200, 250, 300, 350, 400])
y_pred = np.array([210, 240, 320, 330, 415])
````

````python
ss_res = ((y_true - y_pred) ** 2).sum()
ss_tot = ((y_true - y_true.mean()) ** 2).sum()
r2_manual = 1 - ss_res / ss_tot
print(round(r2_manual, 3), round(r2_score(y_true, y_pred), 3))
````

### Ćwiczenie 28
Zaimplementuj **spadek wzdłuż gradientu** dla modelu $\hat{y} = a x + b$ i funkcji kosztu MSE. W każdym kroku:

$$a \leftarrow a - \eta \cdot \frac{2}{n}\sum (\hat{y}_i - y_i)\,x_i \qquad b \leftarrow b - \eta \cdot \frac{2}{n}\sum (\hat{y}_i - y_i)$$

Zacznij od $a = b = 0$, przyjmij $\eta = 0.05$ i wykonaj 2000 iteracji. Wydrukuj `a` i `b` (2 miejsca) oraz wyniki `LinearRegression` - powinny być takie same.

__Oczekiwany wynik:__

```
1.99 1.05
1.99 1.05
```

__Wskazówka:__ Gradient liczymy dla wszystkich obserwacji naraz (tablice NumPy), a parametry aktualizujemy w pętli.

````python
x = np.array([1, 2, 3, 4, 5])
y = np.array([3.1, 4.9, 7.2, 8.8, 11.1])
eta = 0.05
n = len(x)
````

````python
a, b = 0.0, 0.0
for _ in range(2000):
    error = a * x + b - y
    grad_a = 2 / n * (error * x).sum()
    grad_b = 2 / n * error.sum()
    a -= eta * grad_a
    b -= eta * grad_b
print(round(a, 2), round(b, 2))

lr = LinearRegression().fit(x.reshape(-1, 1), y)
print(round(lr.coef_[0], 2), round(lr.intercept_, 2))
````

### Ćwiczenie 29
Wpływ współczynnika uczenia $\eta$. Uruchom spadek wzdłuż gradientu z poprzedniego zadania przez **100 iteracji** dla $\eta \in \{0.001,\ 0.05,\ 0.3\}$ i po zakończeniu wydrukuj dla każdego $\eta$ błąd MSE (format `:.3g`).

Który $\eta$ jest za mały, a który za duży?

__Oczekiwany wynik:__

```
0.001 0.533
0.05 0.0227
0.3 6.35e+158
```

__Wskazówka:__ Zbyt mały $\eta$ - wolne uczenie; zbyt duży - koszt zamiast maleć rośnie (algorytm „rozbiega się”).

````python
x = np.array([1, 2, 3, 4, 5])
y = np.array([3.1, 4.9, 7.2, 8.8, 11.1])
n = len(x)
````

````python
for eta in [0.001, 0.05, 0.3]:
    a, b = 0.0, 0.0
    for _ in range(100):
        error = a * x + b - y
        a -= eta * 2 / n * (error * x).sum()
        b -= eta * 2 / n * error.sum()
    mse = ((a * x + b - y) ** 2).mean()
    print(eta, f'{mse:.3g}')
````

### Ćwiczenie 30
Dane mają zależność **nieliniową** (parabola z szumem). Dla stopni wielomianu 1, 2 i 10 zbuduj potok: `PolynomialFeatures(degree, include_bias=False)` + `LinearRegression`.

Dla każdego stopnia wydrukuj $R^2$ na zbiorze treningowym i testowym (3 miejsca). Który stopień niedouczył model, a który go przeuczył?

__Oczekiwany wynik:__

```
1 0.543 -0.057
2 0.842 0.605
10 0.889 0.316
```

__Wskazówka:__ Niedouczenie: słaby wynik na obu zbiorach. Przeuczenie: bardzo dobry na treningowym, wyraźnie gorszy na testowym.

````python
rng = np.random.RandomState(0)
X = rng.uniform(-3, 3, 60).reshape(-1, 1)
y = 0.5 * X.ravel() ** 2 + X.ravel() + 2 + rng.normal(0, 1, 60)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
````

````python
for degree in [1, 2, 10]:
    model = make_pipeline(PolynomialFeatures(degree, include_bias=False), LinearRegression())
    model.fit(X_train, y_train)
    print(degree, round(model.score(X_train, y_train), 3), round(model.score(X_test, y_test), 3))
````
