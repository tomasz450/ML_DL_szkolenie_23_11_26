### Import bibliotek i dane

````python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer, load_iris
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix,
                             f1_score, precision_score, recall_score, roc_auc_score)
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

dane = load_breast_cancer(as_frame=True)
X = dane.data
y = (dane.target == 0).astype(int)   # 1 = guz złośliwy (klasa pozytywna), 0 = łagodny
print(X.shape, y.value_counts().to_dict())
````

### Ćwiczenie 36
Na podstawie `tn, fp, fn, tp` z macierzy pomyłek oblicz **ręcznie**:

$$\text{precision} = \frac{TP}{TP+FP} \qquad \text{recall} = \frac{TP}{TP+FN} \qquad F_1 = \frac{2 \cdot \text{precision} \cdot \text{recall}}{\text{precision} + \text{recall}}$$

Wydrukuj trzy wartości (3 miejsca), a w drugiej linii te same metryki obliczone funkcjami `precision_score`, `recall_score` i `f1_score`.

__Oczekiwany wynik:__

```
0.98 0.925 0.951
0.98 0.925 0.951
```

__Wskazówka:__ Obie linie powinny dać te same liczby.

````python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=40, stratify=y
)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()
````

````python
precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1 = 2 * precision * recall / (precision + recall)
print(round(precision, 3), round(recall, 3), round(f1, 3))
print(round(precision_score(y_test, y_pred), 3), round(recall_score(y_test, y_pred), 3),
      round(f1_score(y_test, y_pred), 3))
````

### Ćwiczenie 38
Domyślnie klasa 1 jest wskazywana, gdy prawdopodobieństwo $\geq 0.5$. W diagnostyce ważniejsze jest **nie przeoczyć** guza złośliwego.

Wyznacz prawdopodobieństwa klasy 1 (`predict_proba`) i porównaj **recall** oraz **precision** dla progów 0.5 i 0.2. Wydrukuj wyniki w postaci: próg, recall, precision (3 miejsca).

__Oczekiwany wynik:__

```
0.5 0.925 0.98
0.2 0.962 0.911
```

__Wskazówka:__ Obniżenie progu zwykle zwiększa recall kosztem precision.

````python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=40, stratify=y
)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
````

````python
proba = model.predict_proba(X_test)[:, 1]
for threshold in [0.5, 0.2]:
    y_pred = (proba >= threshold).astype(int)
    print(threshold, round(recall_score(y_test, y_pred), 3), round(precision_score(y_test, y_pred), 3))
````

### Ćwiczenie 40
Zbiór `iris` ma **3 klasy** gatunków. Podziel dane (`test_size=0.3`, `random_state=20`, `stratify=y`), wytrenuj `LogisticRegression(max_iter=1000)` i wydrukuj:
- accuracy na zbiorze testowym (4 miejsca),
- macierz pomyłek,
- średnią F1 po klasach (`average='macro'`, 3 miejsca).

__Oczekiwany wynik:__

```
0.9778
[[15  0  0]
 [ 0 14  1]
 [ 0  0 15]]
0.978
```

__Wskazówka:__ W klasyfikacji wieloklasowej metryki liczymy dla każdej klasy osobno i uśredniamy (`average='macro'`).

````python
X_iris, y_iris = load_iris(return_X_y=True)
X_iris.shape, np.bincount(y_iris)
````

````python
X_train, X_test, y_train, y_test = train_test_split(
    X_iris, y_iris, test_size=0.3, random_state=20, stratify=y_iris
)
model = LogisticRegression(max_iter=1000).fit(X_train, y_train)
y_pred = model.predict(X_test)
print(round(accuracy_score(y_test, y_pred), 4))
print(confusion_matrix(y_test, y_pred))
print(round(f1_score(y_test, y_pred, average='macro'), 3))
````
