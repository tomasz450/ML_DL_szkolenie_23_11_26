### Ćwiczenie 9
W tabeli `osoby` zmień nazwy kolumn `'Imię i nazwisko'` na `'imię_nazwisko'` oraz `'Wiek lat'` na `'wiek'`. Następnie ustaw kolumnę `imię_nazwisko` jako indeks i usuń kolumnę `'Miasto'`.

__Oczekiwany wynik:__

```
               wiek
imię_nazwisko      
Anna Nowak       25
Jan Kowalski     41

```

__Wskazówka:__ Użyj `rename(columns=...)`, `set_index()` i `drop(columns=...)`.

````python
osoby = pd.DataFrame({'Imię i nazwisko': ['Anna Nowak', 'Jan Kowalski', 'Ewa Lis'],
                       'Wiek lat': [25, 41, 33],
                       'Miasto': ['Łódź', 'Gdańsk', 'Kraków']})
osoby
````

### Ćwiczenie 10
Tabela `egzaminy` zawiera oceny trzech studentek z trzech egzaminów.

a) Oblicz średnią ocenę z każdego egzaminu (zaokrągloną do 2 miejsc).

b) Dodaj kolumnę `średnia` ze średnią ocen każdej studentki (zaokrągloną do 2 miejsc).

__Oczekiwany wynik:__

```
matematyka     4.0
fizyka         4.0
informatyka    4.0
dtype: float64
```

```
     matematyka  fizyka  informatyka  średnia
Ala           4       3            5      4.0
Ola           5       5            5      5.0
Ela           3       4            2      3.0
```

__Wskazówka:__ Parametr _axis_: `axis=0` liczy po kolumnach (domyślnie), `axis=1` po wierszach.

````python
egzaminy = pd.DataFrame({'matematyka': [4, 5, 3], 'fizyka': [3, 5, 4], 'informatyka': [5, 5, 2]},
                         index=['Ala', 'Ola', 'Ela'])
egzaminy
````

````python
egzaminy.mean()
````

````python
egzaminy['średnia'] = egzaminy.mean(axis=1).round(2)
egzaminy
````
