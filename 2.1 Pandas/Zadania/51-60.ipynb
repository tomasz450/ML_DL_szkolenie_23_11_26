Poniższa komórka przygotowuje oczyszczoną tabelę `auta` (czyszczenie z ćwiczeń 41–50). Uruchom ją bez zmian.

Następnie dla każdej marki policz liczbę ogłoszeń (`liczba`) i średnią cenę (`średnia_cena`). Zostaw marki z co najmniej 50 ogłoszeniami i wyświetl osiem najdroższych (średnia cena zaokrąglona do 2 miejsc).

__Oczekiwany wynik:__


__Wskazówka:__ Użyj `groupby('Marka').agg(...)` z nazwanymi agregacjami, `query()`, `sort_values()` i `head()`.

````python
auta = pd.read_csv('data.csv').drop(columns='Unnamed: 0')
auta['Mileage_wartość'] = pd.to_numeric(auta['Mileage'].str.split().str[0], errors='coerce').replace(0, np.nan)
auta['Engine_cc'] = pd.to_numeric(auta['Engine'].str.replace(' CC', ''), errors='coerce')
auta['Power_bhp'] = pd.to_numeric(auta['Power'].str.replace(' bhp', ''), errors='coerce')
auta['Marka'] = auta['Name'].str.split().str[0].replace({'ISUZU': 'Isuzu', 'Land': 'Land Rover'})
for kolumna in ['Engine_cc', 'Power_bhp']:
    auta[kolumna] = auta[kolumna].fillna(auta.groupby('Marka')[kolumna].transform('median'))
auta['Seats'] = auta['Seats'].fillna(auta['Seats'].median())
auta['wiek'] = 2019 - auta['Year']
auta = auta[auta['Kilometers_Driven'] < 1_000_000]
auta['Właściciele'] = auta['Owner_Type'].map({'First': 1, 'Second': 2, 'Third': 3, 'Fourth & Above': 4})
auta['Automat'] = (auta['Transmission'] == 'Automatic').astype(int)
auta.shape
````
