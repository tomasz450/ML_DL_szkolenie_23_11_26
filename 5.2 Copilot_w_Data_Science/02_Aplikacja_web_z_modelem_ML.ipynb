## 1. Pomysł i prompt dla asystenta

Chcemy aplikację, w której użytkownik ustawia suwakami wymiary kwiatu irysa, a model podaje gatunek wraz z prawdopodobieństwami. Opis zadania dla Copilota (zgodnie z zasadami z poprzedniego notatnika) mógłby brzmieć:

> Zbuduj aplikację w Gradio dla klasyfikatora irysów z pliku `iris_model.joblib` (`RandomForestClassifier`, 4 cechy w cm: długość i szerokość działki kielicha oraz płatka). Model wczytaj raz, przy starcie. Użyj suwaków z sensownymi zakresami i wartościami domyślnymi oraz komponentu `gr.Label`, który pokaże prawdopodobieństwa wszystkich klas. Napisz krótki kod z komentarzami po polsku.
