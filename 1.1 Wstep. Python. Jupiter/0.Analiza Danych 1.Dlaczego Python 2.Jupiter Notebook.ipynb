### Czym jest analiza danych?

Analiza danych to proces przekształcania, oczyszczania i modelowania danych w celu odkrycia użytecznych informacji, wyciągnięcia wniosków oraz wsparcia procesów decyzyjnych. W kontekście szeroko rozumianej nauki o danych, analiza danych stanowi kluczowy element, który łączy różne metody i techniki wykorzystywane do pracy z danymi.

### Główne etapy analizy danych

1. **Zbieranie danych**:
   - Gromadzenie danych z różnych źródeł, takich jak bazy danych, arkusze kalkulacyjne, pliki CSV, API, czujniki IoT, media społecznościowe, itd.
   
2. **Przygotowanie danych**:
   - Oczyszczanie danych (usuwanie lub korygowanie błędów, brakujących wartości).
   - Transformacja danych (normalizacja, standaryzacja, agregacja).
   
3. **Eksploracja danych (Exploratory Data Analysis - EDA)**:
   - Wizualizacja danych (wykresy, histogramy, diagramy pudełkowe).
   - Statystyki opisowe (średnia, mediana, odchylenie standardowe).
   - Wykrywanie wzorców i anomalii.
   
4. **Modelowanie i ocena** (jeśli celem jest predykcja):
   - Wybór i trenowanie modelu (regresja, klasyfikacja, klasteryzacja).
   - Ocena jakości modelu na danych, których model nie widział.
   - Pełny cykl projektu w ujęciu metodologii CRISP-DM omówimy jeszcze dziś.

5. **Prezentacja wyników**:
   - Tworzenie raportów i dashboardów.
   - Interpretacja wyników i wyciąganie wniosków.
   - Prezentacja wyników interesariuszom.

### Techniki i narzędzia

Analiza danych wykorzystuje różnorodne techniki i narzędzia, które pomagają w przekształcaniu surowych danych w wartościowe informacje.

#### Techniki analizy danych

- **Statystyka opisowa**: Podstawowe miary statystyczne, takie jak średnia, mediana, moda, odchylenie standardowe.
- **Wizualizacja danych**: Tworzenie wykresów i diagramów w celu łatwiejszego zrozumienia danych.
- **Modelowanie predykcyjne**: Techniki takie jak regresja, drzewa decyzyjne, lasy losowe, sieci neuronowe.
- **Analiza klasteryzacyjna**: Grupowanie podobnych obiektów w klastry.
- **Analiza koszykowa**: Technika używana w analizie zachowań zakupowych klientów (np. algorytm Apriori).
- **Analiza szeregów czasowych**: Techniki analizy danych zmieniających się w czasie (np. ARIMA, LSTM).

#### Narzędzia analizy danych

- **Jupyter Notebook**: Interaktywne środowisko do kodowania w Pythonie, idealne do eksploracji danych i tworzenia raportów.
- **Pandas**: Biblioteka Pythona do manipulacji i analizy danych strukturalnych.
- **NumPy**: Biblioteka do obliczeń numerycznych w Pythonie.
- **Matplotlib i Seaborn**: Biblioteki do tworzenia wykresów i wizualizacji danych w Pythonie.
- **Scikit-learn**: Biblioteka do uczenia maszynowego w Pythonie.
- **SQL**: Język zapytań używany do zarządzania i manipulacji danymi w bazach danych.

### Zastosowania analizy danych

Analiza danych znajduje zastosowanie w wielu dziedzinach, w tym:

### Interaktywna praca z kodem i danymi

Jupyter Notebook to otwarte środowisko, w którym jeden dokument łączy **kod, wyniki, wykresy i opisy** (Markdown, wzory LaTeX). Notatnik składa się z komórek uruchamianych niezależnie, dzięki czemu pracujemy iteracyjnie: zmiana, uruchomienie, natychmiastowy wynik.

**Dlaczego to ważne w Data Science?**

1. **Natychmiastowa informacja zwrotna** – łatwiej debugować i eksperymentować.
2. **Eksploracja danych (EDA)** – wczytujemy dane, oglądamy je i od razu rysujemy wykresy.
3. **Dokumentacja i prezentacja** – opis, kod i wyniki w jednym miejscu.
4. **Reprodukowalność** – inni mogą powtórzyć analizę (pod warunkiem uruchamiania komórek we właściwej kolejności).

**Typowe zastosowania:** analiza danych (Pandas, NumPy, Matplotlib), uczenie maszynowe (scikit-learn, TensorFlow, PyTorch), nauczanie i raportowanie.

> **Uwaga:** notatnik przechowuje *stan* (zmienne w pamięci kernela), który nie zawsze widać w kodzie. Dlatego tak ważne jest uruchamianie komórek od góry do dołu i regularne `Restart & Run All`.

### Skróty klawiszowe

Notatnik ma dwa tryby: **edycji** (kursor w komórce, wejście: `Enter`) i **komend** (operacje na całych komórkach, wejście: `Esc`).

**Uruchamianie komórek (Jupyter i VS Code)**
- `Shift + Enter` – uruchom komórkę i przejdź do następnej,
- `Ctrl + Enter` – uruchom komórkę i zostań w niej,
- `Alt + Enter` – uruchom komórkę i wstaw nową poniżej.

**Tryb komend (`Esc`)**
- `A` / `B` – wstaw komórkę powyżej / poniżej,
- `M` / `Y` – zmień typ komórki na Markdown / kod,
- `D, D` – usuń komórkę, `Z` – cofnij,
- `C` / `X` / `V` – kopiuj / wytnij / wklej komórkę,
- `Shift + M` – scal zaznaczone komórki *(tylko Jupyter)*,
- `H` – lista wszystkich skrótów *(tylko Jupyter)*.

**Tryb edycji (`Enter`)**
- `Tab` – autouzupełnianie nazw (w VS Code także `Ctrl + Space`),
- `Shift + Tab` – podpowiedź z opisem funkcji (w VS Code: `Ctrl + Shift + Space`),
- `Ctrl + /` – komentowanie / odkomentowanie zaznaczonego kodu (na macOS w VS Code: `Cmd + /`),
- podział komórki w miejscu kursora: w Jupyter `Ctrl + Shift + -`, w VS Code polecenie *Notebook: Split Cell* z palety komend (`Ctrl/Cmd + Shift + P`).

Jeśli nie pamiętasz skrótu, zawsze możesz skorzystać z palety komend (VS Code) lub menu notatnika.

### Sprawdzenie środowiska

Zanim zaczniemy pracę, sprawdzamy, czy interpreter i biblioteki są poprawnie zainstalowane. Komórka poniżej pokazuje wersję Pythona, ścieżkę interpretera oraz wersje pakietów potrzebnych na kolejne dni kursu.

Jeśli widzisz `BRAK` przy pakiecie wymaganym – zainstaluj go w aktywnym środowisku poleceniem `%pip install nazwa-pakietu` (w komórce notatnika) i uruchom komórkę ponownie.

````python
import sys
from importlib.metadata import PackageNotFoundError, version

print("Python     :", sys.version.split()[0])
print("Interpreter:", sys.executable)


def sprawdz(pakiety):
    for pakiet in pakiety:
        try:
            print(f"  OK    {pakiet:<18} {version(pakiet)}")
        except PackageNotFoundError:
            print(f"  BRAK  {pakiet}")


print("\nWymagane (dni 1-4):")
sprawdz(["numpy", "pandas", "matplotlib", "seaborn", "scipy", "scikit-learn", "statsmodels"])

print("\nDodatkowe (kolejne tematy kursu):")
sprawdz(["plotly", "imbalanced-learn", "shap", "lime", "tensorflow", "keras"])
````

### Pierwsza analiza danych

Wykonaj poniższe komórki po kolei. Po każdej odpowiedz na pytanie: co nowego mówi nam wynik?

````python
import pandas as pd

dane = {
    "miasto": ["Warszawa", "Kraków", "Gdańsk", "Wrocław", "Poznań", "Warszawa", "Kraków"],
    "sprzedaz": [120, 95, 80, 90, 70, 130, 105],
    "klienci": [40, 30, 28, 29, 25, 44, 33],
}

df = pd.DataFrame(dane)
df
````

````python
df.describe()
````

````python
df.groupby("miasto", as_index=False)["sprzedaz"].mean().sort_values("sprzedaz", ascending=False)
````

````python
import matplotlib.pyplot as plt

srednia_sprzedaz = df.groupby("miasto")["sprzedaz"].mean().sort_values(ascending=False)
srednia_sprzedaz.plot(kind="bar", title="Średnia sprzedaż wg miasta", figsize=(8, 4))
plt.ylabel("Sprzedaż")
plt.tight_layout()
plt.show();
````

### Najczęstszy błąd na starcie: `NameError`

Najczęstsze przyczyny `NameError`: literówka w nazwie albo nieuruchomiona komórka, w której zmienna jest tworzona (czyli zła kolejność uruchamiania).

````python
import traceback

try:
    dff.head()  # celowy błąd: literówka w nazwie zmiennej
except NameError:
    traceback.print_exc()
````
